## 1.3 Python Semantics<a id="semantics"></a>

### 1.3.1 Indentation
<a id="indentation"></a>
An important concept in data science is that of "scope" of your code or code space: for instance, if you declare a variable, its scope indicates where the variable is present. This leads to the need to organise your code in logical blocks, such as modules, classes, functions, etc...

A very particular aspect of Python is the philosophy of using indentation to deliminate code space. Other languages use braces to delimit their code - both methods of delimitation have their uses: braces are a bit more difficult to parse, while indentations can lead to subtle bugs that are hard to track. 

However, if you use Python, then you don't have a choice: you'll need to use indentation.

For example, this is the code you'd write to create a simple Python loop:

```python

array = np.random.randint(0, 10, 5) # create 5 random numbers between 0 and 10
pivot = 5 # An arbitrary variable of type int
less = [] # An empty list that we will use later
greater = [] # An empty list that we will use later

for x in array:
    if x < pivot:
        less.append(x)
    else:
        greater.append(x)
```

In contrast, this would be a loop in Scala:

```scala
for( a <- 1 to 10){
         val b = a + 10
         println( "Value of a + 10: " + b )}
```
The point I am making is that the Python interpreter knows where the loop statement starts and stops because of the indentation: both the `if` and the `else` statements are indented to the same level, so they are both statements evaluated as part of the loop. Writing the following would result in an error:

```python
for x in array:
    if x < pivot:
        less.append(x)
        else:
            greater.append(x)
```
Because the `else` statement is now executed only if the `if x < pivot:` returns `True`, this else statement will never trigger, even if the initial `if` statement returns `False`. In other words, this `else` statement has no corresponding `if` statement, and attempting to evaluate this code will raise an error. By contrast, with my Scala example, the following would still work fine:

```scala
for( a <- 1 to 10){val b = a + 10
                   if(b < 20){
         println( "Value of a + 10: " + b )}
                               else{return 0}}
```
This is because in Scala, it is the '{}' that determines what is executed in the loop and `if` statement and spacing and lining is here only for clarity of code. The lesson here is that you must pay attention to your whitespaces in Python code; they matter a lot!

You can execute several Python statements on the same line like so:

```python
a = 5; b = 6; c = 7
```

### 1.4.2 Dynamic references, strong types
<a id="dynamicreference"></a>
Python is what is known as a strongly typed language. This means that the interpreter knows at any point what 'type' a variable is. However, Python is a *Dynamic language*, which means it will not use its type information to restrict what you are doing. It makes the language nice to use because of its flexibility. However, this can also be a problem for numerical computation and coding; in contrast, *static typing* (i.e. languages that declare the type of the variable at time of declaration) can catch bugs before execution. Try the following to convince yourself of this issue:

````python
a = 5
type(a)

# a is a integer, i.e. a whole number
````

````python
b='2'
type(b)

# b is a string containing the number 2
````

````python
# Will this code work you think?
a = 5*b
````

Yay! It works! Now let's add 5 to the new value of `a` shall we? Run the code below:

````python
a + 5
````

### So what happened???
Welcome to your first error! As you can see from the message above, you have a 'TypeError' on line 2 because you are trying to concatenate an integer to a string. But what does it mean? Concatenating means putting together. So you are trying to put together a `string` and an `int`.

Take some time to inspect the error message above. You will need to debug your own code a lot, so it is useful to know where to start. I typically read the last line first, that gives me the type of error I am facing. Then I check where the error is, which is indicated above by a dashed green arrow pointing to line 2. So I know that the code that causes the problem is on line 2, `a + 5`. Of course, this is where the error happens ***but not necessarily where the bug is***.

To trace our bug, let's look at what the new variable `a` holds. This might tell us why the code is not running:

````python
# Run this
a
````

As you can see, this result is not what you might have hoped for. Unfortunately, the `string` class in Python has a `*` (multiply) operator. So when we wrote `a = 5*b`, what Python understood is "create a new string with the character '2' (the value of `b`) repeated five times in it". Not at all what we intended. This is because `b` is a `string`, and the fact that the code ran without issues is the effect of strong typing with a dynamic language. So our bug was much, much earlier in the code. It might seem trivial here, but if you have 10'000 lines of code or more, this leads to bugs that are very hard to find and fix.

Let's fix this bug. Of course, we could fix the first line and declare `b` as a `float`. But we can still get the code `a + 5` to run as intended. What we need to do is to 'cast' the string, i.e. transform one of the two data types into another type. In this case, we need to cast `b` into an `int`, which is very simple to do:

````python
a=5 # Let's reset a to its original value
a = a*int(b) # int(b) tries to transform b into an int, which in this case is possible!
a + 5
````

````python
# You can also cast the final 5 to a string for the code to run. Can you guess what the final result would be before you run this cell?

a=5 # Let's reset a to its original value
a = a*b # We have been there before
a + str(5) # Now 5 becomes '5'
````

As you can see, what you do with your datatype plays a big role in the final result. In the first case, you obtain a integer that is multiplication of two integer followed by the sum of one more integer, in the second case you simply obtain a new string that is the sum of 6 substrings (five '2' and one '5').

These types of issues would not occur in programming languages that are statically typed (like Scala) because the type of your variable is declared when you create your variable and cannot be changed later. Because Scala is compiled, I would not even be able to run the code before I fix the type errors: this limits potential bugs later in my code. 

There is one thing you can do with Python, which is relatively new and some people strongly recommend it. You can declare the type of your variable when you create it. For instance:

````python
# See how I 'declare' that my density_calcite variable should be a floating point number here
density_calcite:float = 2.71
````

But at least this way you declare your intention, and it is clear to you (and those reading your code) that this is meant to be a float. 

If you are concerned about the type of a variable, you can query the type of a variable and use control structure to deal with it:

### 1.5.1 if, elif, and else

An **else** statement comes after an **if** or **elif** statement, and executes if the previous conditions are not met but adds no new condition to be met.

````python
# Change the value of seismic_amplitude to see how it impacts the output of the next cell
seismic_amplitude = 4
````

### 1.5.2 for loops<a id="forloop"></a>

A `for loop` is a very important control structure. The basic syntax is this:

```python
for value in collection:
    # do something with aforementioned value
```
    
Later in this class, you will see that, whenever possible, it is preferrable to use a list comprehension rather than a loop. However, this is not always possible or doable, and sometimes a loop is simpler.

````python
sequence = [1, 2, None, 4, None, 5] # In this example, I declare an array named sequence. Two values are 'None' values
total = 0 # I declare a variable named total, that I set to an initial value of 0

# This is a common use of a for loop, to add all values in a collection (my sequence)
# note that I use the values in the sequence:

for value in sequence:
    if value is not None:
        total += value # Only add value to total if it is not 'None'. 'total += value' is equivalent to 
                       # writing 'total = total + value'. There is also a '-=' operator.
total
````

## 1.7 List, Set, and Dictionary (Dict) Comprehensions
<a id="comprehension"></a>
Loops are notoriously error prone, which can lead to subtle bugs that are sometimes difficult to detect. List, set, and dictionary 'comprehensions' are a great way around this problem; we will be using this technique quite a bit in our course. The term 'comprehension' [refers to the ability to create sets or lists based on existing lists/sets](https://en.wikipedia.org/wiki/List_comprehension).

Assuming we have a 'collection' of objects (i.e. either a set, a list, or a dict), we can define a collection comprehension as follows:

```python
new_collection = [val for val in collection if condition]
```

For a dictionary or set, the basic syntax is exactly the same except that you would use curly braces '{}' instead of square braces denoting a list '[]'.

The beauty of this technique is that the entire collection will be accessed without the need for you to use a `for` loop or any other iterative approach. This method is cleaner and avoids many bugs. Let's look at some real examples.


````python
# Let's start with a list of words. We want to capitalise all the words in our first example. 
# We can use a list comprehension for that:

strings = ['a', 'as', 'calcite', 'quartz', 'aragonite', 'magnesite']
[word.upper() for word in strings]
````

Make sure you understand what the code above does, you can think of it as a simple, elegant, less error-prone and faster way to write a loop:
* It takes each value in `strings` one by one, as you would in a loop
* In each loop, it assigns the value of the collection to the variable `word`. This variable name is up to you to decide, but of course you'd want to keep it logical. So, the first loop, `word = 'a'`, while in the second loop `word = 'as'`, etc...
* We then run the method `.upper()` on each `word`, transforming it into the uppercase version of itself
* The transformed words are assigned to the new list that is returned by the function

Note here that I did not save the new list. But I could have easily done so if I had written this:

```python

strings_cap = [word.upper() for word in strings]

```

Let's see how we can use conditionals in list comprehensions:

````python
# Instead let's only keep and capitalise the word if it is longer than 2 characters. 


[word.upper() for word in strings if len(word) > 2]
````

````python
# we can also create a set of unique word lengths using set comprehension and our previously defined collection, 'strings'
unique_lengths = {len(word) for word in strings}
unique_lengths
````

#### Nested list comprehensions

If you have nested lists (i.e. a list of lists) you can still use list comprehension. 

````python
# Let's define a list containing a list of carbonate mineral names and a list of siliciclastic minerals
all_data = [['calcite', 'aragonite', 'dolomite', 'magnesite', 'vaterite'],
            ['quartz', 'kaolinite', 'olivine', 'pyroxene', 'plagioclase']]
````

````python
# Our goal is to keep only those minerals that have more than 9 characters in their names

long_mineral_names = [] # We declare an empty list
for names in all_data: # First for comprehension: this will go through the two lists
    long_enough = [name for name in names if len(name) >= 9] # Second comprehension, this will go through the list of names
    long_mineral_names.extend(long_enough) # We extend our list with the names that are long enough
    
long_mineral_names
````

#### Flattening a list of tuples

We can use the `for` comprehension to 'flatten' a collection. This can be a bit tricky to visualize and is a bit unintuitive. For instance, if we have a list of tuples, we can obtain a list of items using the following approach:

````python
some_tuples = [(1, 2, 3), (4, 5, 6), (7, 8, 9)]
flattened = [x for tup in some_tuples for x in tup]
flattened
````

### 1.8.2 Namespace and Scope of a Variable
<a id="scope"></a>
Two very important concepts in programming are the 'namespace' and 'scope' of a variable. A namespace refers to where your program will search for the name of a variable or a function. The scope of a variable defines the extent in which the variable is available; for example, some variables are only available locally (e.g., within a function), a specific code block (like a loop), or are available globally.

Variables defined in a function are local, and not accessible outside of this function unless they are returned. For instance, the following code will result in an error, try it:

````python
def fun():
    fun_a = 5
    fun_b = 6
    fun_c = 7
    return fun_a, fun_b, fun_c

d, e, f = fun()
fun_a
````

As you can see, Python raised an error because `fun_a` is not defined within the scope of the cell. This is because it is only defined within the scope of the function. What happens if you define a variable in the cell, and try to use it in your function? Let's find out...

````python
x = 'I am declared outside of the function but used inside'
def fun():
    a = 5
    b = 6
    c = 7
    return a, b, c, x

d, e, f, g = fun()
g
````

As you can see, this works because the scope of the variable `x` extends to the function `f()`: both are defined in the same namespace, i.e. the namespace of this entire notebook. In fact, you can call variables defined outside of functions anywhere in the notebook and it will work, even if they are defined further down in your notebook (though they need to be run first, and it can be very confusing).

However, defining a function like I did above is **VERY BAD PRACTICE!** You should never capture global variables within your function and use them, because these variable is not under your direct control and can thus change without you realising it. So it is **much more preferable** to pass the variable you need as an argument to your function. This is how I would refactor (rewrite) my function to be cleaner:

````python
x = 'I am declared outside of the function, then safely passed to the function as an argument and used inside'
def fun(my_variable_x):
    a = 5
    b = 6
    c = 7
    return a, b, c, my_variable_x

d, e, f, g = fun(x)
g
````

The other *huge* benefit of this rewrite is that my code is reusable. Let's say `x` now needs to be 3, I can now reuse the function `f()` very simply:

````python
fun(3)
````

### 1.8.3 Functions Are Objects
<a id="functionsasobjects"></a>
One good thing about Python is that functions are objects, just like anything else in Python. This means that you can pass functions to functions. This ability, often described as 'functions as first-class citizens', means that Python can be used for functional programming. Let's see an example of how this can be used.

````python
# Let's define two functions, one that doubles and one that triples it's input.

def doubles(value):
    return value*2

def triples(value):
    return value*3
````

````python
# Now let's define a function that takes in both a value and a function as arguments. This new function will apply my_function to
# my_value, and add 1 before returning the value

def multiply(my_value, my_function):
    return my_function(my_value) + 1
````

````python
# Now let's use a list of numbers
my_list = [1,2,3,4,5,6]

# We can now define a simple for loop and set a condition to double the even values, but triple the odd values:

for value in my_list:
    the_function = None
    if (value%2 == 0): 
        the_function = doubles
    else: 
        the_function = triples
    result = multiply(value, the_function)
    print(result)
````

The example above is fairly trivial, but you can see how this technique can be useful. First, we only need to write the logic for the multiply function once. This ensures that we don't propagate bugs. Second, it means that you can pass a function as an argument and control the logic of your application in more detail.

### 1.8.4 Anonymous (Lambda) Functions
<a id="lambdas"></a>
An anonymous function, also known as a *lambda* function, is a function whithout a name. Anonymous functions are very helpful for some aspects of functional programming. You can define an anonymous function as:

```python
def short_function(x):
    return x * 2

equiv_anon = lambda x: x * 2
```

Anonymous functions are convenient to use when you want to pass a simple function to another function, but it would be too much overhead to define a named function as you would not use it otherwise. Let's look at an example:

````python
formations = ['Monterey', 'Peedee', 'Shuaiba', 'Nahr Umr', 'Castle Gate']
````

````python
# We can pass an anonymous function that uses the last letter of the formation name to sort the list.
formations.sort(key=lambda x: x[-1])
formations
````

Make sure you understand the logic of the code above: we use a `lambda` function that returns the last character of a string (`x[-1]`: check the section about slicing if you are not comfortable with this notation) and this is the character that will be used for sorting. So *Shuaiba* is the first formation name in the new list because it ends in 'a', etc...

This type of approach gives you a lot of power to do complex data manipulation easily. 

### 1.8.5 Currying: Partial Argument Application
<a id="lambdas"></a>
The term 'Currying' is another functional programming term. It refers to partially applying a function that has multiple arguments to create a new function that requires less arguments. For instance, we can create a function that adds two numbers:

````python
def add_numbers(x, y):
    return x + y
````

If we wanted to create a new function that would always use the value '5' as the first number, we can curry the 'add_number' function and create an 'add_five' function:

````python
# That is one way to do it, using an anonymous function
add_five = lambda y: add_numbers(5, y)
````

````python
# We can also import the functional tools and do a partial function:
from functools import partial
add_five = partial(add_numbers, 5)
````

Either way, we now have a new, curried function that adds five to whatever we give it:

````python
add_five(2)
````

## 1.9 The Art of Writing Clean Code
<a id="cleancode"></a>

One important part is writing logical and unique variable names. I could have called the density of calcite **'x'** above, and the total weight **'y'**. But these are not meaningful variable names. I could have written **'dcal'** and **'twcal'** for these variables. However, these are not as intuitive and make your code much harder to read. Avoid this and use reasonable variable names. Don't go over the top either. For instance, **'density_of_calcite_in_grams_per_cubic_centimeter'** might be clear but it is too long and will clutter your code. Find a good compromise between readability and intuitive names.
