### Understanding the 'pip' command
*This is an optional read only if you are interested - feel free to skip it*

The commands above are used to install dependencies with a Python package manager. The standard manager is called `pip` and it relies on the `pypi` repository. Most Python packages are stored in `pypi`. Anaconda Python comes with its own package manager called `conda`. `conda` works like `pip` and handles most of the packages (includings the ones in bonth the `conda` and the `pypi` repositories). You should always use `conda` first when you have Anaconda Python, and you cannot use `conda` if you have the standard distribution of Python.

The command above is not a Python command, but a terminal command. We can use the `!` operator in any Python cell to tell Jupyter that the command we want to run is in fact a terminal command (i.e. system command). So if you were to run this from a terminal (not straight from here) you would use:

`conda install -r requirements.txt`

Let's dissect this command:

* `!`: This magic commands tells Jupyter to run the code as a system command
* `conda` (or `pip`): this is the program we will use. The other elements are `arguments` to the program.
* `install`: pretty self explanatory. You could also `uninstall` and do several other things with `conda`/`pip`
* `-r`: this argument (or flag) is here to indicate that we want to install dependencies *recursively*, i.e. more than just one program. This is generally followed by a filename that contains the dependencies. In our case, the file is called `requirements.txt`. If you open it you will see that I listed a number of dependencies that I want to install for you.
* Alternatively, you can install only one dependency at a time. For instance running `conda install scikit-learn` would only install `scikit-learn`

## Module and Class imports in Scikit-learn
We will rely on the libary `Scikit-learn` (aka `sklearn`) a lot during this class. Scikit-learn is a vast libary, and so learning the different modules and submodules is a challenge in itself. There are many ways to import modules and classes in notebooks, but there is a best practice, certainly for `sklearn`!

### DON'Ts:

🚫 Import of entire library <br>
<code>import sklearn

**Must type library and module prefix every time**
